### 9. Random Module

```python
# np.random.rand(10) generates 10 random decimal numbers between 0 and 1.
np.random.rand(10)
```

```python
# np.random.randn(10) generates 10 random numbers from a standard normal distribution (mean = 0, standard deviation = 1).
np.random.randn(10)
```

```python
# np.random.randint(1,10,10) generates 10 random integers from 1 to 9 (10 is excluded).
np.random.randint(1,50,10)
```
