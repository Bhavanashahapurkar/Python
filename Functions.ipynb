### USER DEFINE FUNCTION

```python
# A user-defined function is a function created by the programmer to perform a specific task.
# It is created using the def keyword.
```

```python
def add(a,b):
    return a+b
```

```python
add(22,3)
```

```python
def add(a,b):
    print(a+b)
```

```python
add(1,2)
```

```python
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
def div(a,b):
    return a/b
```

```python
add(22,3)
```

```python
sub(2,1)
```

```python
mul(2,3)
```

```python
div(2,2)
```

```python
def palindrome(name):
    name=name.lower()
    return name==name[::-1]
    
    
```

```python
palindrome("Pop")
```

```python
def number(a):
    if (a%2==0):
      return "Even Number"
    else:
       return "Odd Number"
```

```python
number(4)
```

```python
number(3)
```

```python
def number(a):
    if (a>0):
      return "Positive Number"
    else:
       return "Negative Number"
```

```python
number(12)
```

```python
number(-1)
```

```python
number(0)
```
